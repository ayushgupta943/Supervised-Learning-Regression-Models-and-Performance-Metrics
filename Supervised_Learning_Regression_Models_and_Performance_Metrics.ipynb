1. What is Simple Linear Regression (SLR)? Explain its purpose.

  - Simple Linear Regression (SLR) is a method used to study the relationship  between two variables — one independent (X) and one dependent (Y).
   
    It fits a straight line to predict Y based on X.

Equation:
        
    Y=b0​+b1​X

Purpose:

- To predict one variable using another.

- To understand the relationship between variables.

- To analyze trends in data.

Example: Predicting exam marks (Y) from hours studied (X).

2. What are the key assumptions of Simple Linear Regression?

  - The key assumptions of Simple Linear Regression are:

    - Linearity:
    The relationship between X and Y is linear.

    - Independence:
    The observations are independent of each other.

    - Homoscedasticity:
    The variance of errors is constant across all values of X.

    - Normality:
    The errors (residuals) are normally distributed.

    - No Multicollinearity:
    (Not relevant for SLR, but for MLR) — since there’s only one X variable.

In short:
The data should have a linear, independent, constant variance, and normally distributed error pattern.

3. Write the mathematical equation for a simple linear regression model and
explain each term.

 - The mathematical equation of a Simple Linear Regression model is:

        Y=b0​+b1​X+ε

    Where:

    Y → Dependent variable (the value we want to predict)

    X → Independent variable (the predictor)

    b₀ → Intercept (value of Y when X = 0)

    b₁ → Slope (change in Y for a one-unit change in X)

    ε (epsilon) → Error term (difference between actual and predicted Y)

In short:
It represents a straight-line relationship between X and Y.

6. What is Logistic Regression? How does it differ from Linear Regression?

 - Logistic Regression is a statistical method used to predict a categorical   (usually binary) outcome, such as Yes/No, Pass/Fail, or Spam/Not Spam, based on one or more independent variables.

   It uses the logistic (sigmoid) function to model the probability that a given input belongs to a particular class.

    
    P(Y=1)=1+e−(b0​+b1​X)1​

Difference between Linear and Logistic Regression

  - Type of Output:

    Linear Regression → Predicts continuous values

    Logistic Regression → Predicts categorical outcomes (0 or 1)

  - Output Range:

    Linear → Any real number

    Logistic → Between 0 and 1

  - Type of Relationship:

    Linear → Fits a straight line

    Logistic → Fits an S-shaped (sigmoid) curve

 -  Use Cases:

    Linear → Predict marks, prices, weight, etc.

    Logistic → Predict pass/fail, yes/no, disease/no disease

 - Error Measurement:

    Linear → Measured using Mean Squared Error (MSE)

    Logistic → Measured using Log Loss (cross-entropy)


7. Name and briefly describe three common evaluation metrics for regression
models.

  - Mean Absolute Error (MAE):

    Measures the average absolute difference between actual and predicted values.

    Lower MAE indicates better model accuracy.

  - Mean Squared Error (MSE):

    Measures the average squared difference between actual and predicted values.

    Penalizes larger errors more than MAE.

  - R-squared (R²):

    Represents the proportion of variance in the dependent variable explained by the model.

    Value ranges from 0 to 1; closer to 1 means better fit.


8. What is the purpose of the R-squared metric in regression analysis?

  - The R-squared (R²) metric measures how well a regression model explains the variability of the dependent variable.

    It shows the proportion of variance in the dependent variable that is explained by the independent variable(s).

    Value range: 0 to 1

    Closer to 1 → model explains most of the variance (better fit)

    Closer to 0 → model explains very little of the variance (poor fit)



```python
#9. Python code to fit a Simple Linear Regression model using scikit-learn
# Import necessary libraries
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
# X = independent variable (e.g., hours studied)
# Y = dependent variable (e.g., exam score)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([45, 50, 55, 60, 65])

# Create the linear regression model
model = LinearRegression()

# Fit the model to the data
model.fit(X, Y)

# Print the slope (b1) and intercept (b0)
print("Slope (b1):", model.coef_[0])
print("Intercept (b0):", model.intercept_)
```

10. How do you interpret the coefficients in a Simple Linear Regression model?

  - In a simple linear regression model:
  
        Y=b0​+b1​X

    Intercept (b₀):

     - Represents the predicted value of Y when X = 0.

     - Example: If b₀ = 40, then when X = 0, Y = 40.

    Slope (b₁):

     - Represents the change in Y for a one-unit increase in X.

     - Example: If b₁ = 5, then for every additional unit of X, Y increases by 5 units.
